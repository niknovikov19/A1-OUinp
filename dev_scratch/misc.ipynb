```python
import matplotlib.pyplot as plt
import numpy as np

f = 3
T = 1 / f
tmax = 1 - T / 2
t = np.linspace(0, tmax, 2000)
x = np.sin(2 * np.pi * 3 * t)

plt.figure(figsize=(8, 1.6))
plt.plot(t, x, '.-', linewidth=4, markersize=12)
plt.tick_params(labelbottom=False, labelleft=False, bottom=False, left=False)
plt.axis('off')
plt.ylim(-1.8, 1.8)
plt.show()
```

```python
import matplotlib.colors as mcolors

t = np.linspace(0, tmax, 5000)

col = plt.rcParams['axes.prop_cycle'].by_key()['color'][0]
col = np.array(mcolors.to_rgb(col))
N = 4

plt.figure(figsize=(8, 1.6))
for n in range(N):
    k = n / (N - 1)
    x = np.sin(2 * np.pi * (3 * t + k))
    col_ = col * k + np.array([1, 1, 1]) * (1 - k)
    plt.plot(t, x, '.-', linewidth=2, markersize=6, color=col_)
plt.tick_params(labelbottom=False, labelleft=False, bottom=False, left=False)
plt.axis('off')
plt.ylim(-1.8, 1.8)
plt.show()
```
